```python
import json

def load_rows(path):
    with open(path, encoding="utf-8") as f:
        text = f.read()
    return json.loads(text[text.index("["): text.rindex("]") + 1])

female = load_rows("../tests/fixtures/FemalePlayerCareer.js")
male = load_rows("../tests/fixtures/MalePlayerCareer.js")

for i, (f_val, m_val) in enumerate(zip(female[0], male[0])):
    print(i, repr(f_val), "|", repr(m_val))
```
