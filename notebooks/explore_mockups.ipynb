```python
import json

def load_rows(path):
    with open(path, encoding="utf-8") as f:
        text = f.read()
    return json.loads(text[text.index("["): text.rindex("]") + 1])

female = load_rows("../tests/fixtures/FemalePlayerCareer.js")
male = load_rows("../tests/fixtures/MalePlayerCareer.js")

for i, (f_val, m_val) in enumerate(zip(female[0], male[0])):
    print(i, repr(f_val), "|", repr(m_val))
```

```python
import re

def _parse_matches(js_text):
    """Extract the list of match records from a Tennis Abstract career JS file."""
    found = re.search(r"morematchmx\s*=\s*(\[\[.*?\]\])", js_text, re.DOTALL)
    if found is None:
        raise ValueError("No match data (morematchmx) found in the JS text.")
    return json.loads(found.group(1))
```

```python
with open("../tests/fixtures/FemalePlayerCareer.js", encoding="utf-8") as f:
    female_text = f.read()
with open("../tests/fixtures/MalePlayerCareer.js", encoding="utf-8") as f:
    male_text = f.read()

female = _parse_matches(female_text)
male = _parse_matches(male_text)

print(len(female), len(male))          # expect: 3 3
print(female[0][11], "|", male[2][11]) # expect: Anne Miller | Israel Matos Gil
print(len(female[0]), len(male[0])) 
```
